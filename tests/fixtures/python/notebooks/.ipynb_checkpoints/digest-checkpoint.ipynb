# Release digest
Pulls the week's releases and summarizes them.

```python
!pip install requests openai
%matplotlib inline
```

```python
import os
import requests
from openai import OpenAI

TRACKER = "https://api.tracker-hub.dev/v2"
client = OpenAI(api_key=os.environ["OPENAI_API_KEY"])
```

```python
def releases(team):
    resp = requests.get(
        f"{TRACKER}/teams/{team}/releases",
        headers={"Authorization": f"Bearer {os.environ['TRACKER_TOKEN']}"},
        params={"since": "7d"},
    )
    return resp.json()
```

```python
%%time
notes = releases("platform")
summary = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": str(notes)}],
)
```
